# Subspace alignment

```python
npcs = analysis_params['npcs']
nmin_units = analysis_params['nunits_column']
niterations = analysis_params['niterations']
nsites = len(column_units)
population_results = {align_event: {} for align_event in align_events}
```

```python
# Organize trials and concatenate
bin_fs = 1/postproc_params['bin_width']
neural_idx_start = lda_metadata['trelevant1_smooth']
neural_idx_stop = lda_metadata['trelevant2_smooth']
unique_targets = np.unique(fr_zscore['MOVEMENT ONSET']['target'][0])
ntargets = len(unique_targets)

np.random.seed(0)
shuffled_trials_idx = np.arange(ds_params['min_trials_to_target']*len(unique_targets))
np.random.shuffle(shuffled_trials_idx)

conc_neural_data = {}
conc_target_data = {}
conc_pos_data = {}
conc_velo_data = {}

for align_event in align_events:
    conc_neural_data[align_event] = []
    conc_pos_data[align_event] = []
    conc_velo_data[align_event] = []
    conc_target_data[align_event] = []
    for irec in tqdm(range(nrecs)):
        if ~np.isnan(fr_zscore[align_event]['align_spikes_zscore_smooth'][irec]).all():
            smoothed_neural_data = fr_zscore[align_event]['align_spikes_zscore_smooth'][irec]

            # Organize trials
            trials = []
            for itarget, target in enumerate(unique_targets):
                trials.extend(np.where(fr_zscore[align_event]['target'][irec] == target)[0]) # Randomly sample trials from across acceptable trials within session
            trials = np.array(trials)
            trials = trials[shuffled_trials_idx]

            prep_data = []
            target_data = []
            for trial_idx in trials:
                target_data.extend([fr_zscore[align_event]['target'][irec][trial_idx]] * (neural_idx_stop-neural_idx_start))
                prep_data.extend(smoothed_neural_data[neural_idx_start:neural_idx_stop,:,trial_idx])

            prep_data = np.array(prep_data)
            conc_neural_data[align_event].append(prep_data)
            conc_target_data[align_event].append(np.array(target_data))
            
        else:
            conc_neural_data[align_event].append([])
            conc_target_data[align_event].append([])
```

```python
conc_random_prep_data = []
for irec in tqdm(range(nrecs)):
    smoothed_neural_data = fr_zscore['RANDOM']['align_spikes_zscore_smooth'][irec]
    if ~np.isnan(fr_zscore[align_event]['align_spikes_zscore_smooth'][irec]).all():
        ntrials = smoothed_neural_data.shape[1]
        random_neural_data_temp = []
        for trial_idx in range(ntrials):
            random_neural_data_temp.extend(smoothed_neural_data[neural_idx_start:neural_idx_stop,trial_idx,:])
        conc_random_prep_data.append(np.array(random_neural_data_temp))
    else:
        conc_random_prep_data.append([])       

intertrial_array = np.concatenate([conc_random_prep_data[irec] for irec in range(len(conc_random_prep_data)) if len(conc_random_prep_data[irec])>0], axis=1)
```

```python
## Look at spatial distribution of units
unit_pos = []
for ipenetration in tqdm(range(nrecs)):
    nunits_site = len(unit_df[unit_df['penetration']==ipenetration])
    unit_pos.extend([(unit_df[unit_df['penetration']==ipenetration]['rec_rcaxis'].iloc[0], unit_df[unit_df['penetration']==ipenetration]['rec_mlaxis'].iloc[0]) for iunit in range(nunits_site)])
        
unit_pos = np.array(unit_pos)
for align_event in align_events:
    population_results[align_event]['unit_pos'] = unit_pos
```

```python
conc_neural_data_all_units = {}
for align_event in align_events:
    all_prep_data_temp = [conc_neural_data[align_event][irec] for irec in range(nrecs) if len(conc_neural_data[align_event][irec]) > 0]
    conc_neural_data_all_units[align_event] = np.concatenate(all_prep_data_temp, axis=1)
```

## Split by recording site

```python
nt = neural_idx_stop - neural_idx_start
for align_event in align_events:
    population_results[align_event]['column'] = {}
    population_results[align_event]['column']['cca'] = np.zeros((nsites, nsites, npcs))*np.nan
    population_results[align_event]['column']['cca_sd'] = np.zeros((nsites, nsites, npcs))*np.nan
    population_results[align_event]['column']['aligned_decoding'] = np.zeros((nsites, nsites))*np.nan
    for isite in tqdm(range(nsites)):
        for isite2 in range(nsites):
            if len(column_units[isite]) >= nmin_units and len(column_units[isite2]) >= nmin_units:
                temp_ccs = np.zeros((niterations, npcs))*np.nan
                temp_decoding = np.zeros((niterations))*np.nan
                for iiter in range(niterations):
                    np.random.seed(iiter)
                    random_unit_idx1 = np.random.choice(np.array(column_units[isite]['index']), nmin_units, replace=False)
                    if isite == isite2:
                        random_unit_idx2=random_unit_idx1
                    else:
                        random_unit_idx2 = np.random.choice(np.array(column_units[isite2]['index']), nmin_units, replace=False)
                    
                    proj_data1 = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx1]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx1],axis=0))
                    proj_data2 = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx2]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx2],axis=0))                
                    
                    Ma, Mb, S, U, V = canoncorr_gallego(proj_data1[:,:npcs], proj_data2[:,:npcs], fullReturn=True)
                    temp_ccs[iiter,:] = S

                    trial_seperated_U = np.zeros((nt, ds_params['min_trials_to_target']*ntargets, npcs))*np.nan
                    trial_seperated_V = np.zeros((nt, ds_params['min_trials_to_target']*ntargets, npcs))*np.nan
                    trial_seperated_target = np.zeros((ds_params['min_trials_to_target']*ntargets))*np.nan
                    for itrial in range((ds_params['min_trials_to_target']*ntargets)):
                        start_idx = itrial*nt
                        stop_idx = start_idx+nt
                        trial_seperated_U[:,itrial,:] = U[start_idx:stop_idx,:]
                        trial_seperated_V[:,itrial,:] = V[start_idx:stop_idx,:]
                        trial_seperated_target[itrial] = conc_target_data[align_event][isite][start_idx+1]
                    
                    # Train decoder using baseline site dynamics projected onto the shared dimensions
                    lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
                    lda.fit(np.mean(trial_seperated_U,axis=0), trial_seperated_target)
                    temp_decoding[iiter] = lda.score(np.mean(trial_seperated_V,axis=0), trial_seperated_target)
                    
                population_results[align_event]['column']['cca'][isite,isite2,:] = np.mean(temp_ccs, axis=0)
                population_results[align_event]['column']['cca_sd'][isite,isite2,:] = np.std(temp_ccs, axis=0)
                population_results[align_event]['column']['aligned_decoding'][isite,isite2] = np.mean(temp_decoding)
```

```python
# Organize data by column decoding accuracy for plotting
for align_event in align_events:
    column_rc_pos = np.array([column_units[irec]['rec_rcaxis'][0] for irec in range(len(column_units))])
    column_ml_pos = np.array([column_units[irec]['rec_mlaxis'][0] for irec in range(len(column_units))])
    decoding = np.array([np.mean(lda_results[align_event]['pseudopopulations']['column_scores'][isite]) for isite in range(nsites)])
    good_nan_mask = ~np.isnan(population_results[align_event]['column']['cca'][:,0,0])
    cr_sort = np.argsort(decoding[good_nan_mask])
    input_array = np.mean(population_results[align_event]['column']['cca'][good_nan_mask,:,:][:,good_nan_mask,:][cr_sort,:,:][:,cr_sort,:analysis_params['nmodes_avg_cca']], axis=2)
    population_results[align_event]['column']['cca_acc_ordered'] = input_array
    population_results[align_event]['column']['aligned_decoding_acc_ordered'] = population_results[align_event]['column']['aligned_decoding'][good_nan_mask,:][:,good_nan_mask][cr_sort,:][:,cr_sort]
    population_results[align_event]['column']['acc_ordered'] = 100*decoding[good_nan_mask][cr_sort]
```

```python
# Organize data by column rostral-caudal position for plotting
for align_event in align_events:
    column_rc_pos = np.array([column_units[irec]['rec_rcaxis'][0] for irec in range(len(column_units))])
    column_ml_pos = np.array([column_units[irec]['rec_mlaxis'][0] for irec in range(len(column_units))])
    good_nan_mask =  ~np.isnan(population_results[align_event]['column']['cca'][:,0,0])
    cr_sort = np.argsort(column_rc_pos[good_nan_mask])
    input_array = np.mean(population_results[align_event]['column']['cca'][good_nan_mask,:,:][:,good_nan_mask,:][cr_sort,:,:][:,cr_sort,:analysis_params['nmodes_avg_cca']], axis=2)
    population_results[align_event]['column']['cca_rc_ordered'] = input_array
    population_results[align_event]['column']['aligned_decoding_rc_ordered'] = population_results[align_event]['column']['aligned_decoding'][good_nan_mask,:][:,good_nan_mask][cr_sort,:][:,cr_sort]
    rc_neighboring_ccs = get_neighboring_ccs(input_array, nneighbors=analysis_params['nneighbors'])
    population_results[align_event]['column']['rc_neighboring_cca'] = rc_neighboring_ccs
    population_results[align_event]['column']['rc_order'] = column_rc_pos[good_nan_mask][cr_sort]
    population_results[align_event]['column']['ml_order'] = column_ml_pos[good_nan_mask][cr_sort]
```

```python
# Compute cortical surface distance between each recording site
for align_event in align_events:
    pos_diff = np.zeros((nsites, nsites))*np.nan
    for isite1 in range(nsites):
        for isite2 in range(nsites):
            if isite1 != isite2:
                site1_pos = np.array((column_units[isite1]['rec_rcaxis'].iloc[0], column_units[isite1]['rec_mlaxis'].iloc[0]))
                site2_pos = np.array((column_units[isite2]['rec_rcaxis'].iloc[0], column_units[isite2]['rec_mlaxis'].iloc[0]))
                pos_diff[isite1,isite2] = np.sqrt(np.sum((site1_pos - site2_pos)**2))
    population_results[align_event]['column']['pos_diff'] = pos_diff
```

```python
# Compute neighboring CCA correlation when organized by decoding accurcy
for align_event in align_events:
    column_decoding = np.array([np.mean(lda_results[align_event]['pseudopopulations']['column_scores'][irec]) for irec in range(nsites)])
    good_nan_idx = np.where(np.logical_and(np.sum(np.isnan(population_results[align_event]['column']['cca'][:,:,0]), axis=0) < nsites, ~np.isnan(column_decoding)))[0]
    np_acc_ordered_site_idx = good_nan_idx[np.argsort(column_decoding[good_nan_idx])]
    np_acc_neighboring_ccs = get_neighboring_ccs(np.mean(population_results[align_event]['column']['cca'][np_acc_ordered_site_idx,:,:][:,np_acc_ordered_site_idx,:analysis_params['nmodes_avg_cca']],axis=2), nneighbors=analysis_params['nneighbors'])
    population_results[align_event]['column']['acc_neighboring_cca'] = np_acc_neighboring_ccs
```

```python
# Compute neighboring CCA correlation when organized by various different axis.
column_positions = np.array([(column_units[irec]['rec_rcaxis'].iloc[0], column_units[irec]['rec_mlaxis'].iloc[0]) for irec in range(len(column_units))])
for align_event in align_events:
    population_results[align_event]['column']['random_r2'] = np.zeros((analysis_params['niterations']))*np.nan
    population_results[align_event]['column']['random_angles'] = np.zeros((analysis_params['niterations']))*np.nan
        
    good_nan_idx = np.where(np.sum(np.isnan(population_results[align_event]['column']['cca'][:,:,0]), axis=0) < nsites)[0]

    xmin, xmax = -5,5
    # Random axis 
    for iaxis in tqdm(range(analysis_params['niterations'])):
        np.random.seed(iaxis)
        random_angle = np.random.rand(1)*np.pi*2
        population_results[align_event]['column']['random_angles'][iaxis] = random_angle[0]
        random_axis = column_positions @ np.array([[np.cos(random_angle), -np.sin(random_angle)],[np.sin(random_angle), np.cos(random_angle)]])[:,:,0]
        any_ordered_site_idx = good_nan_idx[np.argsort(random_axis[good_nan_idx,0])]
        any_neighboring_ccs = get_neighboring_ccs(np.mean(population_results[align_event]['column']['cca'][any_ordered_site_idx,:,:][:,any_ordered_site_idx,:analysis_params['nmodes_avg_cca']], axis=2), nneighbors=analysis_params['nneighbors'])

        fit, fitscore, pcc, pval, regfit = aopy.analysis.base.linear_fit_analysis2D(random_axis[any_ordered_site_idx,0], any_neighboring_ccs)
        population_results[align_event]['column']['random_r2'][iaxis] = fitscore
        intercept, slope = regfit.intercept_, regfit.coef_[0]
```

## All units

```python
# Plot some trajectories
align_event = align_events[-1]
population_results[align_events[-1]]['example'] = {}
nunits_all_split = analysis_params['nunits_cca']
ntrials = ds_params['min_trials_to_target']*ntargets
nt = neural_idx_stop - neural_idx_start
trial_structured_La = np.zeros((nt,npcs,ntrials))*np.nan
trial_structured_Lb = np.zeros((nt,npcs,ntrials))*np.nan
trial_structured_U = np.zeros((nt,npcs,ntrials))*np.nan
trial_structured_V = np.zeros((nt,npcs,ntrials))*np.nan
trial_structured_cca_proj_data1 = np.zeros((nt,npcs,ntargets))*np.nan
trial_structured_cca_proj_data2 = np.zeros((nt,npcs,ntargets))*np.nan
trial_separated_target = np.zeros((ntrials))*np.nan

# np.random.seed(4) #
# random_unit_idx = np.random.choice(np.arange(conc_neural_data_all_units[align_event].shape[1]), (nunits_all_split,2), replace=False)
# La = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx[:,0]]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx[:,0]],axis=0))[:,:npcs]
# Lb = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx[:,1]]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx[:,1]],axis=0))[:,:npcs]

np.random.seed(1) #
neural_data_a = conc_neural_data[align_event][analysis_params['example_site_cca']['recs'][0]]
neural_data_b = conc_neural_data[align_event][analysis_params['example_site_cca']['recs'][1]]
nunits_temp = np.min([neural_data_a.shape[1], neural_data_b.shape[1]])
random_unit_idx1 = np.random.choice(np.arange(neural_data_a.shape[1]), nunits_temp, replace=False)
random_unit_idx2 = np.random.choice(np.arange(neural_data_b.shape[1]), nunits_temp, replace=False)
La = PCA(svd_solver='full').fit_transform(neural_data_a[:,random_unit_idx1]/np.std(neural_data_a[:,random_unit_idx1],axis=0))[:,:npcs]
Lb = PCA(svd_solver='full').fit_transform(neural_data_b[:,random_unit_idx2]/np.std(neural_data_b[:,random_unit_idx2],axis=0))[:,:npcs]
# La = PCA(svd_solver='full').fit_transform(neural_data_a[:,:]/np.std(neural_data_a[:,:],axis=0))[:,:npcs]
# Lb = PCA(svd_solver='full').fit_transform(neural_data_b[:,:]/np.std(neural_data_b[:,:],axis=0))[:,:npcs]

Ma, Mb, S, U, V = canoncorr_gallego(La, Lb, fullReturn=True)

for itrial in range((ntrials)):
    start_idx = itrial*nt
    stop_idx = start_idx+nt
    trial_structured_La[:,:,itrial] = La[start_idx:stop_idx,:]
    trial_structured_Lb[:,:,itrial] = Lb[start_idx:stop_idx,:]
    trial_structured_U[:,:,itrial] = U[start_idx:stop_idx,:]
    trial_structured_V[:,:,itrial] = V[start_idx:stop_idx,:]
    trial_separated_target[itrial] = conc_target_data[align_event][0][start_idx+1]

population_results[align_events[-1]]['example']['La'] = np.zeros((nt, npcs, ntargets))*np.nan
population_results[align_events[-1]]['example']['Lb'] = np.zeros((nt, npcs, ntargets))*np.nan
    
U, _, Vh = scipy.linalg.svd(Ma, full_matrices=False, compute_uv=True, overwrite_a=False, check_finite=False)
for itarget in range(ntargets):
    target_idx_use = trial_separated_target==(itarget+1)
    trial_structured_cca_proj_data1[:,:,itarget] = np.mean(trial_structured_La[:,:,target_idx_use], axis=2) @ U @ Vh
    population_results[align_events[-1]]['example']['La'][:,:,itarget] = np.mean(trial_structured_La[:, :,target_idx_use], axis=2)
    population_results[align_events[-1]]['example']['Lb'][:,:,itarget] = np.mean(trial_structured_Lb[:, :,target_idx_use], axis=2)

U, _, Vh = scipy.linalg.svd(Mb, full_matrices=False, compute_uv=True, overwrite_a=False, check_finite=False)
for itarget in range(ntargets):
    target_idx_use = trial_separated_target==(itarget+1)
    trial_structured_cca_proj_data2[:,:,itarget] = np.mean(trial_structured_Lb[:,:,target_idx_use], axis=2) @ U @ Vh
    
# Compute aligned decoding
lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
lda.fit(np.mean(trial_structured_U,axis=0).T, trial_separated_target)
population_results[align_events[-1]]['example']['aligned_decoding'] = lda.score(np.mean(trial_structured_V,axis=0).T, trial_separated_target)
population_results[align_events[-1]]['example']['CCAa'] = trial_structured_cca_proj_data1
population_results[align_events[-1]]['example']['CCAb'] = trial_structured_cca_proj_data2
```

```python
# Look at random groups of units. 
nunits_all_split = analysis_params['nunits_cca']
ntrials = ds_params['min_trials_to_target']*ntargets
nt = neural_idx_stop - neural_idx_start
for align_event in align_events:
    population_results[align_event]['random'] = {}
    population_results[align_event]['random']['within_acc_La'] = np.zeros((analysis_params['niterations']))*np.nan
    population_results[align_event]['random']['within_acc_Lb'] = np.zeros((analysis_params['niterations']))*np.nan
    population_results[align_event]['random']['cca'] = np.zeros((analysis_params['niterations'],npcs))*np.nan #random_units_S
    population_results[align_event]['random']['across_acc_La'] = np.zeros((analysis_params['niterations']))*np.nan 

    for iiter in tqdm(range(analysis_params['niterations'])):
        trial_structured_La = np.zeros((nt,ntrials,npcs))*np.nan
        trial_structured_Lb = np.zeros((nt,ntrials,npcs))*np.nan
        trial_structured_U = np.zeros((nt,ntrials,npcs))*np.nan
        trial_structured_V = np.zeros((nt,ntrials,npcs))*np.nan
        trial_separated_target = np.zeros((ntrials))*np.nan
        np.random.seed(iiter)
        random_unit_idx = np.random.choice(np.arange(conc_neural_data_all_units[align_event].shape[1]), (analysis_params['nunits_cca'],2), replace=False)

        La = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx[:,0]]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx[:,0]],axis=0))[:,:npcs]
        Lb = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,random_unit_idx[:,1]]/np.std(conc_neural_data_all_units[align_event][:,random_unit_idx[:,1]],axis=0))[:,:npcs]

        Ma, Mb, S, U, V = canoncorr_gallego(La, Lb, fullReturn=True)
        population_results[align_event]['random']['cca'][iiter,:] = S
        # population_results[align_event]['random']['Ma'].append(Ma)
        # population_results[align_event]['random']['Mb'].append(Mb)
        for itrial in range((ntrials)):
            start_idx = itrial*nt
            stop_idx = start_idx+nt
            trial_structured_La[:,itrial,:] = La[start_idx:stop_idx,:]
            trial_structured_Lb[:,itrial,:] = Lb[start_idx:stop_idx,:]
            trial_structured_U[:,itrial,:] = U[start_idx:stop_idx,:]
            trial_structured_V[:,itrial,:] = V[start_idx:stop_idx,:]
            trial_separated_target[itrial] = conc_target_data[align_event][0][start_idx+1]

        # population_results[align_event]['random']['La'].append(trial_structured_La)
        # population_results[align_event]['random']['Lb'].append(trial_structured_Lb)
        # Get within La data
        lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
        lda.fit(np.mean(trial_structured_La,axis=0),trial_separated_target)
        population_results[align_event]['random']['within_acc_La'][iiter] = lda.score(np.mean(trial_structured_La,axis=0), trial_separated_target)

        # Get within Lb data
        lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
        lda.fit(np.mean(trial_structured_Lb,axis=0),trial_separated_target)
        population_results[align_event]['random']['within_acc_Lb'][iiter] = lda.score(np.mean(trial_structured_Lb,axis=0), trial_separated_target)

        # Get across La, Lb score
        lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
        lda.fit(np.mean(trial_structured_U,axis=0),trial_separated_target)
        population_results[align_event]['random']['across_acc_La'][iiter] = lda.score(np.mean(trial_structured_V,axis=0), trial_separated_target)
```

## Use Gallego approach to split units into tuned and untuned

```python
align_event = align_events[-1]
ntargets = len(unique_targets)
ntrials = 320
mean_fr_per_cond = []
for irec in range(nrecs):
    if len(prep_data_all[align_event][irec]) > 0:
        mean_fr_per_cond_temp = np.zeros((prep_data_all[align_event][irec].shape[1], ntargets))*np.nan
        for itarg in range(ntargets):
            temp_target_idx = target_data_all[align_event][irec] == (itarg+1)
            mean_fr_per_cond_temp[:,itarg] = np.mean(prep_data_all[align_event][irec][temp_target_idx,:], axis=0)
        mean_fr_per_cond.append(mean_fr_per_cond_temp)
mean_fr_per_cond = np.concatenate(mean_fr_per_cond)
```

```python
# Compute tuning
thetas = np.arange(ntargets)*45
fit_params, md, pd = aopy.analysis.tuning.run_tuningcurve_fit(mean_fr_per_cond, thetas)
```

```python
# Compute R2 between fit curve and actual data
tuning_r2 = np.zeros(mean_fr_per_cond.shape[0])*np.nan
for iunit in range(mean_fr_per_cond.shape[0]):
    fit_curve = fit_params[iunit,0]*np.cos(thetas) + fit_params[iunit,1]*np.sin(thetas) + fit_params[iunit, 2]
    tuning_r2[iunit] = sklearn.metrics.r2_score(mean_fr_per_cond[iunit,:], fit_curve)
```

```python
r2_thresh = 0.2
fig, ax = plt.subplots(1,1,figsize=(4,3))
ax.plot(tuning_r2[np.argsort(tuning_r2)])
ax.plot(ax.get_xlim(), [r2_thresh, r2_thresh], 'r--')
ax.set(xlabel='Unit', ylabel='Tuning \ncurve r2')
fig.tight_layout()
plt.show()
```

```python
# Compute CCs the same way as Gallego paper
proj_data_all = {}
proj_data_untuned = {}
proj_data_tuned = {}

# for align_event in align_events:
all_prep_data = []
for irec in range(nrecs):
    if len(prep_data_all[align_event][irec]) > 0:
        all_prep_data.append(prep_data_all[align_event][irec])
all_prep_data = np.concatenate(all_prep_data, axis=1)

unit_mask_tuned = tuning_r2 > r2_thresh
unit_mask_untuned = np.argsort(tuning_r2)[:np.sum(unit_mask_tuned)]

proj_data_all[align_event] = PCA(svd_solver='full').fit_transform(all_prep_data[:,:]/np.std(all_prep_data[:,:],axis=0))
proj_data_tuned[align_event] = PCA(svd_solver='full').fit_transform(all_prep_data[:,unit_mask_tuned]/np.std(all_prep_data[:,unit_mask_tuned],axis=0))
proj_data_untuned[align_event] = PCA(svd_solver='full').fit_transform(all_prep_data[:,unit_mask_untuned]/np.std(all_prep_data[:,unit_mask_untuned],axis=0))
```

```python
ccs_pp_tuning = canoncorr_gallego(proj_data_tuned[align_event][:,:npcs], proj_data_untuned[align_event][:,:npcs])
ccs_pp_tuning_tunedvall = canoncorr_gallego(proj_data_tuned[align_event][:,:npcs], proj_data_all[align_event][:,:npcs])
ccs_pp_tuning_untunedvall = canoncorr_gallego(proj_data_untuned[align_event][:,:npcs], proj_data_all[align_event][:,:npcs])
```

```python
# Repeat for different groups of units
nunits_per_group = 50
prep_onto_prep_groups_var_ratio = np.zeros((len(list(high_dec_groups.keys())), npcs, niterations))*np.nan
prep_onto_move_groups_var_ratio = np.zeros((len(list(high_dec_groups.keys())), npcs, niterations))*np.nan
move_onto_prep_groups_var_ratio = np.zeros((len(list(high_dec_groups.keys())), npcs, niterations))*np.nan
move_onto_move_groups_var_ratio = np.zeros((len(list(high_dec_groups.keys())), npcs, niterations))*np.nan
for igroup, group_label in enumerate(tqdm(list(high_dec_groups.keys()))):
    for iiter in range(niterations):
        unit_idx = np.random.choice(np.where(high_dec_groups[group_label])[0], nunits_per_group, replace=False)
        
        prep_tavg_group = []
        move_tavg_group = []
        prep_tavg_group_array = np.concatenate([tavg_prep_data[prep_event][irec] for irec in range(len(tavg_prep_data[prep_event])) if len(tavg_prep_data[prep_event][irec])>0], axis=2)[:,:,unit_idx]
        move_tavg_group_array = np.concatenate([tavg_prep_data[move_event][irec] for irec in range(len(tavg_prep_data[move_event])) if len(tavg_prep_data[move_event][irec])>0], axis=2)[:,:,unit_idx]
        prep_tavg_group_array = (prep_tavg_group_array - np.mean(prep_tavg_group_array,axis=1)[:,None,:])
        move_tavg_group_array = (move_tavg_group_array - np.mean(move_tavg_group_array,axis=1)[:,None,:])
        for itarget in range(ntargets):
            prep_tavg_group.extend(prep_tavg_group_array[:,itarget,:])
            move_tavg_group.extend(move_tavg_group_array[:,itarget,:])
        prep_tavg_group = np.array(prep_tavg_group)
        move_tavg_group = np.array(move_tavg_group)
        
        prep_pca = PCA(svd_solver='full').fit(prep_tavg_group/np.std(prep_tavg_group,axis=0)) # Use only top 2 prep pcs (from Elsayed paper)
        move_pca = PCA(svd_solver='full').fit(move_tavg_group/np.std(move_tavg_group,axis=0)) # Use only top 4 move pcs (From Elsayed paper)

        prep_onto_prep_groups_var_ratio[igroup, :, iiter] = 100*np.var(prep_pca.transform(prep_tavg_group), axis=0)[:npcs]/np.sum(np.var(prep_pca.transform(prep_tavg_group), axis=0))
        prep_onto_move_groups_var_ratio[igroup, :, iiter] = 100*np.var(move_pca.transform(prep_tavg_group), axis=0)[:npcs]/np.sum(np.var(move_pca.transform(prep_tavg_group), axis=0))
        move_onto_prep_groups_var_ratio[igroup, :, iiter] = 100*np.var(prep_pca.transform(move_tavg_group), axis=0)[:npcs]/np.sum(np.var(prep_pca.transform(move_tavg_group), axis=0))
        move_onto_move_groups_var_ratio[igroup, :,iiter] = 100*np.var(move_pca.transform(move_tavg_group), axis=0)[:npcs]/np.sum(np.var(move_pca.transform(move_tavg_group), axis=0))
```

```python
# Start by splitting out high and low decoding units and decoding across them without alignment
accross_highlow_accuracy_pca = {}
for align_event in align_events:
    low_group_max = np.max(lda_results[align_event]['single_ch_decoding_all'][np.argsort(lda_results[align_event]['single_ch_decoding_all'])][:nunits_per_group_split])
    high_group_min = np.min(lda_results[align_event]['single_ch_decoding_all'][np.argsort(lda_results[align_event]['single_ch_decoding_all'])][-nunits_per_group_split:])
    accross_highlow_accuracy_pca[align_event] = []
    for iiter in tqdm(range(analysis_params['niterations'])):
        np.random.seed(iiter)
        low_unit_idx = np.random.choice(np.where(lda_results[align_event]['single_ch_decoding_all'] <= low_group_max)[0], (analysis_params['nunits_cca'],2), replace=False)
        high_unit_idx = np.random.choice(np.where(lda_results[align_event]['single_ch_decoding_all'] >= high_group_min)[0], (analysis_params['nunits_cca'],2), replace=False)
        unit_idx_interactions = [low_unit_idx[:,0], low_unit_idx[:,1], high_unit_idx[:,0], high_unit_idx[:,1]]
        
        temp_score = np.zeros((len(unit_idx_interactions), len(unit_idx_interactions)))*np.nan
        for train_site, train_unit_idx in enumerate(unit_idx_interactions):
            for test_site, test_unit_idx in enumerate(unit_idx_interactions):
                proj_data1 = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,train_unit_idx]/np.std(conc_neural_data_all_units[align_event][:,train_unit_idx],axis=0))
                proj_data2 = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,test_unit_idx]/np.std(conc_neural_data_all_units[align_event][:,test_unit_idx],axis=0))                

                # Separate aligned data into trials before decoding
                trial_seperated_pd1 = np.zeros((nt, ds_params['min_trials_to_target']*8, npcs))*np.nan
                trial_seperated_pd2 = np.zeros((nt, ds_params['min_trials_to_target']*8, npcs))*np.nan
                trial_seperated_target = np.zeros((ds_params['min_trials_to_target']*8))*np.nan
                for itrial in range((ds_params['min_trials_to_target']*8)):
                    start_idx = itrial*nt
                    stop_idx = start_idx+nt
                    trial_seperated_pd1[:,itrial,:] = proj_data1[start_idx:stop_idx,:npcs]
                    trial_seperated_pd2[:,itrial,:] = proj_data2[start_idx:stop_idx,:npcs]
                    trial_seperated_target[itrial] = conc_target_data[align_event][0][start_idx]

                lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
                lda.fit(np.mean(trial_seperated_pd1, axis=0), trial_seperated_target)
                temp_score[train_site, test_site] = lda.score(np.mean(trial_seperated_pd2, axis=0), trial_seperated_target)

        accross_highlow_accuracy_pca[align_event].append(temp_score)
        
        
```

#### Split by task relevance (prep vs. move vs. shared) groups

```python
# Compute CCs the same way as Gallego paper
nunits_per_group = 30
task_event_cca_decoding = {}
for align_event in align_events:
    task_event_cca_decoding[align_event] = []
    
    for iiter in tqdm(range(niterations)):
        temp_proj_data = {}
        for igroup, group in enumerate(list(high_dec_groups.keys())):
            unit_idx = np.random.choice(np.where(high_dec_groups[group])[0], nunits_per_group, replace=False)
            temp_proj_data[group] = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event][:,unit_idx]/np.std(conc_neural_data_all_units[align_event][:,unit_idx],axis=0))

        compiled_cca_decoding = np.zeros((len(list(high_dec_groups.keys())), len(list(high_dec_groups.keys()))))*np.nan
        for itraingroup, traingroup in enumerate(list(high_dec_groups.keys())):
            for itestgroup, testgroup in enumerate(list(high_dec_groups.keys())):
                La = temp_proj_data[traingroup][:,:npcs]
                Lb = temp_proj_data[testgroup][:,:npcs]

                Ma, Mb, S, U, V = canoncorr_gallego(La, Lb, fullReturn=True)
                lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
                lda.fit(U, conc_target_data[align_event][0])

                compiled_cca_decoding[itraingroup, itestgroup] = lda.score(V,conc_target_data[align_event][0])


        task_event_cca_decoding[align_event].append(compiled_cca_decoding)
```

```python
### Now align activity from each even for each group to another (across events)
across_groups_and_events_La = {}
across_groups_and_events_Lb = {}
across_groups_and_events_aligned_decoding = np.zeros((len(list(high_dec_groups.keys())), len(align_events), len(align_events), niterations))*np.nan
for idec_group, decoding_group in enumerate(tqdm(list(high_dec_groups.keys()))):
    across_groups_and_events_La[decoding_group] = []
    across_groups_and_events_Lb[decoding_group] = [] 
    for ievent_train, align_event_train in enumerate(align_events):
        for ievent_test, align_event_test in enumerate(align_events):
            for iiter in range(niterations):
                
                unit_idx = np.random.choice(np.where(high_dec_groups[decoding_group])[0], nunits_per_group, replace=False)
                
                train_temp_proj_data = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event_train][:,unit_idx])
                test_temp_proj_data = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event_test][:,unit_idx])
                
                # Perform align dynamic decoding (Gallego 2018)
                La = train_temp_proj_data[:,:npcs]
                Lb = test_temp_proj_data[:,:npcs]

                Ma, Mb, S, U, V = canoncorr_gallego(La, Lb, fullReturn=True)
                lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
                lda.fit(U, conc_target_data[align_event][0])

                across_groups_and_events_aligned_decoding[idec_group, ievent_train, ievent_test, iiter] = lda.score(V,conc_target_data[align_event][0])                
                
            across_groups_and_events_La[decoding_group].append(La)
            across_groups_and_events_Lb[decoding_group].append(Lb)
```

```python
# Repeat but with randomly selected units
across_groups_and_events_aligned_decoding_random = np.zeros((len(align_events), len(align_events), niterations))*np.nan
for ievent_train, align_event_train in enumerate(tqdm(align_events)):
    for ievent_test, align_event_test in enumerate(align_events):
        for iiter in range(niterations):
                # Perform align dynamic decoding on randomly selected units
                random_unit_idx = np.random.choice(np.arange(conc_neural_data_all_units[align_event].shape[1]), nunits_per_group, replace=False)
                
                random_train_temp_proj_data = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event_train][:,random_unit_idx]/np.std(conc_neural_data_all_units[align_event_train][:,random_unit_idx],axis=0))
                random_test_temp_proj_data = PCA(svd_solver='full').fit_transform(conc_neural_data_all_units[align_event_test][:,random_unit_idx]/np.std(conc_neural_data_all_units[align_event_test][:,random_unit_idx],axis=0))
                
                # Perform align dynamic decoding (Gallego 2018)
                La = random_train_temp_proj_data[:,:npcs]
                Lb = random_test_temp_proj_data[:,:npcs]

                Ma, Mb, S, U, V = canoncorr_gallego(La, Lb, fullReturn=True)
                lda = sklearn.discriminant_analysis.LinearDiscriminantAnalysis()
                lda.fit(U, conc_target_data[align_event][0])

                across_groups_and_events_aligned_decoding_random[ievent_train, ievent_test, iiter] = lda.score(V,conc_target_data[align_event][0])   
```

```python
prep_event = align_events[0]
move_event = align_events[2]
prep_tavg = []
move_tavg = []
prep_tavg_array = np.concatenate([tavg_prep_data[prep_event][irec] for irec in range(len(tavg_prep_data[prep_event])) if len(tavg_prep_data[prep_event][irec])>0], axis=2)[:,:,:]
move_tavg_array = np.concatenate([tavg_prep_data[move_event][irec] for irec in range(len(tavg_prep_data[move_event])) if len(tavg_prep_data[move_event][irec])>0], axis=2)[:,:,:]
prep_tavg_array = (prep_tavg_array - np.mean(prep_tavg_array,axis=1)[:,None,:])
move_tavg_array = (move_tavg_array - np.mean(move_tavg_array,axis=1)[:,None,:])
for itarget in range(ntargets):
    prep_tavg.extend(prep_tavg_array[:,itarget,:])
    move_tavg.extend(move_tavg_array[:,itarget,:])
prep_tavg = np.array(prep_tavg)
move_tavg = np.array(move_tavg)
prep_move_data = np.concatenate((prep_tavg, move_tavg))

prep_pca = PCA(svd_solver='full').fit(prep_tavg)
move_pca = PCA(svd_solver='full').fit(move_tavg)
```

```python
# Repeat for different groups of units
nunits_per_group = 30
prep_onto_move_r2_groups = np.zeros((len(list(high_dec_groups.keys())), niterations))*np.nan
move_onto_prep_r2_groups = np.zeros((len(list(high_dec_groups.keys())), niterations))*np.nan
for igroup, group_label in enumerate(tqdm(list(high_dec_groups.keys()))):
    for iiter in range(niterations):
        unit_idx = np.random.choice(np.where(high_dec_groups[group_label])[0], nunits_per_group, replace=False)
        
        prep_tavg_group = []
        move_tavg_group = []
        prep_tavg_group_array = np.concatenate([tavg_prep_data[prep_event][irec] for irec in range(len(tavg_prep_data[prep_event])) if len(tavg_prep_data[prep_event][irec])>0], axis=2)[:,:,unit_idx]
        move_tavg_group_array = np.concatenate([tavg_prep_data[move_event][irec] for irec in range(len(tavg_prep_data[move_event])) if len(tavg_prep_data[move_event][irec])>0], axis=2)[:,:,unit_idx]
        prep_tavg_group_array = (prep_tavg_group_array - np.mean(prep_tavg_group_array,axis=1)[:,None,:])
        move_tavg_group_array = (move_tavg_group_array - np.mean(move_tavg_group_array,axis=1)[:,None,:])
        for itarget in range(ntargets):
            prep_tavg_group.extend(prep_tavg_group_array[:,itarget,:])
            move_tavg_group.extend(move_tavg_group_array[:,itarget,:])
        prep_tavg_group = np.array(prep_tavg_group)
        move_tavg_group = np.array(move_tavg_group)
        
        prep_pca_group = PCA(svd_solver='full').fit(prep_tavg_group) # Use only top 2 prep pcs (from Elsayed paper)
        move_pca_group = PCA(svd_solver='full').fit(move_tavg_group) # Use only top 4 move pcs (From Elsayed paper)

        Xprep = prep_pca_group.transform(prep_tavg_group_array[prep_idx_compre,:,:])[:,:2] # Use only top 2 prep pcs (from Elsayed paper)
        Xmove = move_pca_group.transform(move_tavg_group_array[move_idx_compare,:,:])[:,:4] # Use only top 4 move pcs (From Elsayed paper)
        
        W, _ = weiner_filter(Xprep, Xmove, lags=0, fit_intercept=False) # Projects prep onto move
        prep_onto_move_r2_groups[igroup, iiter] = 1 - np.linalg.norm(Xmove - (Xprep @ W)) / np.linalg.norm(Xmove)
        move_onto_prep_r2_groups[igroup, iiter] = 1 - np.linalg.norm(Xprep - (Xmove @ np.linalg.pinv(W))) / np.linalg.norm(Xprep)
```

```python
fig, ax = plt.subplots(1,2,figsize=(10,4))
ax[0].plot(np.mean(prep_onto_move_r2_groups, axis=1))
ax[0].fill_between(np.arange(4), np.mean(prep_onto_move_r2_groups, axis=1)-np.std(prep_onto_move_r2_groups, axis=1), np.mean(prep_onto_move_r2_groups, axis=1)+np.std(prep_onto_move_r2_groups, axis=1), alpha=0.5)
ax[1].plot(np.mean(move_onto_prep_r2_groups, axis=1))
ax[1].fill_between(np.arange(4), np.mean(move_onto_prep_r2_groups, axis=1)-np.std(move_onto_prep_r2_groups, axis=1), np.mean(move_onto_prep_r2_groups, axis=1)+np.std(move_onto_prep_r2_groups, axis=1), alpha=0.5)
ax[0].set(xlabel='Unit Group', ylabel='Trajectory $R^2$', title='Prep onto Move', ylim=(0.3, 0.7))
ax[1].set(xlabel='Unit Group', ylabel='Trajectory $R^2$',  title='Move onto Prep')
ax[0].set_xticks(np.arange(4), ['N.M ','P','M', 'P+M'], fontsize=18), ax[1].set_xticks(np.arange(4), ['N.M ','P','M', 'P+M'], fontsize=18)
fig.tight_layout()
plt.show()
```
