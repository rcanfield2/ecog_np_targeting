This script does the following:
1. Registers neuropixel insert hole locations to ECoG electrodes in space
2. Loads ECoG decoding map
3. Interpolates to estimate the ECoG decoding accuracy at each neuropixel insert hole
4. Saves interpolated results.

```python
import numpy as np
import matplotlib.pyplot as plt
import aopy
import os
import pandas as pd
from db import dbfunctions as db
from ipywidgets import interactive, widgets
import scipy
import h5py
from tqdm.auto import tqdm 
import seaborn as sn
import sklearn
from sklearn.decomposition import PCA, FactorAnalysis
from itertools import compress
import multiprocessing as mp
import time
import math
from scipy.fft import fft
```

```python
from local_functions import load_yaml_params
subject = 'affi'
param_filename = 'spatiotemporal_structure_params.yaml'
ds_params, data_paths, save_paths, filenames, postproc_params, uq_params, analysis_params, vis_params, target_colors = load_yaml_params(param_filename)
```

```python
subjects= ds_params['subjects']
np_inserts = ds_params['implant_names']
ninserts = len(np_inserts)
save_figs = True
```

# Register neuropixel insert hole locations to ECoG electrodes in space

```python
np_ecog_mapping = {}
for isubject, subject in enumerate(subjects):    
    np_ecog_mapping[subject] = {}
    np_ecog_mapping[subject]['ecog'] = {}

    # Load ecog and neuropixel drive information and align to ecog
    for iinsert, insert in enumerate(np_inserts):  
        np_ecog_mapping[subject][insert] = {}
        np_pos, ecog_pos, np_number, ecog_acq_elecs = aopy.data.base.align_neuropixel_recoring_drive(insert, 'ECoG244', subject, theta=postproc_params['drive_alignment_theta'][subject])
        np_ecog_mapping[subject][insert]['pos'] = np.round(np_pos,4)
        np_ecog_mapping[subject][insert]['number'] = np_number
        np_ecog_mapping[subject]['ecog']['elec_pos'] = np.round(ecog_pos,3)
        np_ecog_mapping[subject]['ecog']['acq_ch'] = ecog_acq_elecs
        np_ecog_mapping[subject]['ecog']['elec_channels'] = np.array(list(np.arange(1,len(ecog_pos)+1)), dtype='float')
```

```python
subject = subjects[0]
for iinsert, insert in enumerate(np_inserts):
    fig, ax = plt.subplots(1,2,figsize=(10,3))
    data_map = aopy.visualization.get_data_map(np_ecog_mapping[subject]['ecog']['elec_channels'], np_ecog_mapping[subject]['ecog']['elec_pos'][:,0], np_ecog_mapping[subject]['ecog']['elec_pos'][:,1])
    im = aopy.visualization.plot_spatial_map(data_map, np_ecog_mapping[subject]['ecog']['elec_pos'][:,0], np_ecog_mapping[subject]['ecog']['elec_pos'][:,1], cmap='YlGnBu',nan_color='white', ax=ax[0])
    ax[0].plot(np_ecog_mapping[subject][insert]['pos'][:,0], np_ecog_mapping[subject][insert]['pos'][:,1], 'r.', label=f"{insert}")
    ax[0].set(xlim=(-6.5,6.5), ylim=(-6.5,6.5))
    cb = plt.colorbar(im, label='ECoG Electrode Number')

    im = aopy.visualization.plot_spatial_map(data_map, np_ecog_mapping[subject]['ecog']['elec_pos'][:,0], np_ecog_mapping[subject]['ecog']['elec_pos'][:,1], cmap='YlGnBu', nan_color='white', ax=ax[1])
    [ax[1].annotate(str(np_ecog_mapping[subject][insert]['number'][ipt]), (np_ecog_mapping[subject][insert]['pos'][ipt,0], np_ecog_mapping[subject][insert]['pos'][ipt,1]), ha='center', va='center', color='r') for ipt in range(np_ecog_mapping[subject][insert]['pos'].shape[0])]
    ax[1].set(xlim=(-6.5,6.5), ylim=(-6.5,6.5))
    plt.suptitle(insert)
    fig.tight_layout()
    plt.show()
```

```python
for subject in subjects:
    for insert in np_inserts:
        # np_ecog_mapping[subject][f"{insert}_nloc"] = np.sum(~np.isnan(np_ecog_mapping[subject][insert]['pos'][:,0]))
        data_map = aopy.visualization.get_data_map(np_ecog_mapping[subject]['ecog']['elec_channels'], np_ecog_mapping[subject]['ecog']['elec_pos'][:,0], np_ecog_mapping[subject]['ecog']['elec_pos'][:,1])
        data_map_ecogch, data_map_ecogch_xy = aopy.visualization.calc_data_map(np_ecog_mapping[subject]['ecog']['elec_channels'], np_ecog_mapping[subject]['ecog']['elec_pos'][:,0], np_ecog_mapping[subject]['ecog']['elec_pos'][:,1], (len(np.unique(np_ecog_mapping[subject]['ecog']['elec_pos'][:,0])), len(np.unique(np_ecog_mapping[subject]['ecog']['elec_pos'][:,1]))), interp_method='linear')
        np_ecog_mapping[subject][f"{insert}_interp"] = scipy.interpolate.interpn((np.unique(np_ecog_mapping[subject]['ecog']['elec_pos'][:,0]),np.unique(np_ecog_mapping[subject]['ecog']['elec_pos'][:,1])), data_map_ecogch.T, np_ecog_mapping[subject][insert]['pos'], method='linear', bounds_error=False)
```

```python
for subject in subjects:
    fig, ax = plt.subplots(1,ninserts,figsize=(ninserts*4,3))
    for iinsert, insert in enumerate(np_inserts):
        npinsert_interp_datamap = aopy.visualization.get_data_map(np_ecog_mapping[subject][f"{insert}_interp"], np_ecog_mapping[subject][insert]['pos'][:,0], np_ecog_mapping[subject][insert]['pos'][:,1])
        im = aopy.visualization.plot_spatial_map(npinsert_interp_datamap, np_ecog_mapping[subject][insert]['pos'][:,0], np_ecog_mapping[subject][insert]['pos'][:,1], cmap='YlGnBu', ax=ax[iinsert], nan_color='white')
        cb = plt.colorbar(im, label='ECoG Elec Num')
        ax[iinsert].set_title(f"{insert} - {subject}")

    fig.tight_layout()
    plt.show()
```

# Load decoding accuracy map 

```python
# Load decoding accuracy and plot
dec_map_results = {}
fig, ax = plt.subplots(1,2,figsize=(8,3))
for isubject, subject in enumerate(subjects):
    dec_map_results[subject] = {}    
    dec_map_results[subject]['ecog_decoding_accuracy'] = np.load(os.path.join(save_paths['ecog_maps'], f"{subject}_movement_LinRegAll_hgamma.npz")) # High gamma linear regression
    dec_map_results[subject]['ecog_dec_acc_elec_pos'] = np.round(dec_map_results[subject]['ecog_decoding_accuracy']['elec_pos'], 8)
    dec_map_results[subject]['ecog_decoding_accuracy'] = dec_map_results[subject]['ecog_decoding_accuracy']['da']
    # else: 
    #     dec_map_results[subject]['ecog_decoding_accuracy'] = np.load(os.path.join(save_paths['ecog_maps'], f"{subject}_movement_LinRegAll_hgamma.npz")) # High gamma linear regression
    #     dec_map_results[subject]['ecog_dec_acc_elec_pos'] = np.round(dec_map_results[subject]['ecog_decoding_accuracy']['elec_pos'], 8)
    #     dec_map_results[subject]['ecog_decoding_accuracy'] = dec_map_results[subject]['ecog_decoding_accuracy']['da']
    
    # Center electrode positions
    dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0] = dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0] - np.median(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0]))
    dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1] = dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1] - np.median(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1]))
    
    dec_map_results[subject]['data_map_ecogch'], dec_map_results[subject]['data_map_ecogch_xy'] = aopy.visualization.calc_data_map(dec_map_results[subject]['ecog_decoding_accuracy'], dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0], dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1], (len(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0])), len(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1]))), interp_method='linear')
    im = aopy.visualization.plot_spatial_map(dec_map_results[subject]['data_map_ecogch'], dec_map_results[subject]['data_map_ecogch_xy'][0], dec_map_results[subject]['data_map_ecogch_xy'][1], cmap='YlGnBu', ax=ax[isubject], nan_color='white')
    cb = plt.colorbar(im)
    
    ax[isubject].set(title=f"{subject} - ECoG Decoding Accuracy")

fig.tight_layout()
if save_figs:
    aopy.visualization.base.savefig(save_paths['ecog_maps'], 'ecog_decoding_accuracy_maps.svg')
```

```python
import matplotlib
aopy.utils.release_memory_limit()
cmap_name = 'cool'
site_numbers = {}
site_insert_name = {}
for subject in subjects:
    if subject == 'beignet':
        _, _, preproc_metadata = aopy.data.base.pkl_read(f"{subject}_{filenames['preprocessed']}", save_paths['postproc_data'])
        print(preproc_metadata.keys())
        site_numbers[subject] = preproc_metadata['recording_sites'] -1
        site_insert_name[subject] = ['NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 'NP_Insert72', 
                         'NP_Insert137', 'NP_Insert137', 'NP_Insert137', 'NP_Insert137', 'NP_Insert137', 'NP_Insert137', 'NP_Insert137']
    elif subject2plot == 'affi':
        _, _,  preproc_metadata = aopy.data.base.pkl_read(f"{subject}_{filenames['preprocessed']}", save_paths['postproc_data'])
        site_numbers[subject] = preproc_metadata['recording_sites'] -1
        site_insert_name[subject] = preproc_metadata['implant']
```

```python
for subject in subjects:
    dec_map_results[subject]['rec_sites'] = site_numbers[subject]
    dec_map_results[subject]['rec_locations'] = np.array([[np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,0][site], np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,1][site]] for isite, site in enumerate(site_numbers[subject])])
    if subject == 'affi':
        dec_map_results[subject]['rc_axis_angle'] = np.deg2rad(postproc_params['rc_axis_theta'][subject]) # rotates 'x-axis' by this much
        dec_map_results[subject]['rc_axis'] = dec_map_results[subject]['rec_locations'] @ np.array([[np.cos(dec_map_results[subject]['rc_axis_angle']), -np.sin(dec_map_results[subject]['rc_axis_angle'])],[np.sin(dec_map_results[subject]['rc_axis_angle']), np.cos(dec_map_results[subject]['rc_axis_angle'])]])
        dec_map_results[subject]['rc_axis'][:,0] += postproc_params['rc_chamber_offset'][subject]
    elif subject == 'beignet':
        dec_map_results[subject]['rc_axis_angle'] = np.deg2rad(postproc_params['rc_axis_theta'][subject]) # rotates 'x-axis' by this much
        dec_map_results[subject]['rc_axis'] = dec_map_results[subject]['rec_locations'] @ np.array([[np.cos(dec_map_results[subject]['rc_axis_angle']), -np.sin(dec_map_results[subject]['rc_axis_angle'])],[np.sin(dec_map_results[subject]['rc_axis_angle']), np.cos(dec_map_results[subject]['rc_axis_angle'])]])
        dec_map_results[subject]['rc_axis'][:,0] += postproc_params['rc_chamber_offset'][subject]
    
    # map_min = np.min(np.max(dec_map_results[subject]['rc_axis'][:,0]))
    map_min = np.min(dec_map_results[subject]['rc_axis'][:,0])
    map_min_val = 0.25
    map_range = np.max(dec_map_results[subject]['rc_axis'][:,0]) - np.min(dec_map_results[subject]['rc_axis'][:,0])
    day_colors = [matplotlib.colormaps[cmap_name](((dec_map_results[subject]['rc_axis'][irec,0]+np.abs(map_min))/map_range)*(1-map_min_val)+map_min_val)[:3] for irec in range(dec_map_results[subject]['rc_axis'].shape[0])]
    dec_map_results[subject]['day_colors']=day_colors
```

```python
# Repeat plots with dots at specified lcoations
site_label_color= 'r'
# Plot decoding accuracy
for subject in subjects:
    fig, ax = plt.subplots(1,3,figsize=(12,3))
    dec_map_results[subject]['data_map_ecogch'], dec_map_results[subject]['data_map_ecogch_xy'] = aopy.visualization.calc_data_map(dec_map_results[subject]['ecog_decoding_accuracy'], dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0], dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1], (len(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,0])), len(np.unique(dec_map_results[subject]['ecog_dec_acc_elec_pos'][:,1]))), interp_method='linear')
    im = aopy.visualization.plot_spatial_map(dec_map_results[subject]['data_map_ecogch'], dec_map_results[subject]['data_map_ecogch_xy'][0], dec_map_results[subject]['data_map_ecogch_xy'][1], cmap='Greys', ax=ax[0], nan_color='white')
    cb = plt.colorbar(im)
    
    # Label relevant dots
    [ax[0].plot(np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,0][site], np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,1][site], '.', color=dec_map_results[subject]['day_colors'][isite], markersize=14) for isite, site in enumerate(site_numbers[subject])]
    ax[0].set(title=f"{subject} - ECoG Decoding Accuracy")
    
    ax[0].set(xlim=(-6,6), ylim=(-6,6))

    im = aopy.visualization.plot_spatial_map(dec_map_results[subject]['data_map_ecogch'], dec_map_results[subject]['data_map_ecogch_xy'][0], dec_map_results[subject]['data_map_ecogch_xy'][1], cmap='YlGnBu', ax=ax[1], nan_color='white')
    cb = plt.colorbar(im)

    # Label relevant dots
    [ax[1].annotate(f"{int(site_numbers[subject][isite]+1)}", (np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,0][site], np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,1][site]),ha='center', va='center', ) for isite, site in enumerate(site_numbers[subject])]
    ax[1].set(title=f"{subject} - ECoG Decoding Accuracy")
    ax[1].set(xlim=(-6,6), ylim=(-6,6))

    # Label relevant dots
    [ax[2].plot(np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,0][site], np_ecog_mapping[subject][site_insert_name[subject][isite]]['pos'][:,1][site], '.', color=dec_map_results[subject]['day_colors'][isite], markersize=14) for isite, site in enumerate(site_numbers[subject])]
    im = aopy.visualization.plot_spatial_map(dec_map_results[subject]['data_map_ecogch']/np.nanmax(dec_map_results[subject]['data_map_ecogch']), dec_map_results[subject]['data_map_ecogch_xy'][0], dec_map_results[subject]['data_map_ecogch_xy'][1], cmap='YlGnBu', ax=ax[2], nan_color='white')
    cb = plt.colorbar(im, label='$\mu$ECoG Reponse [a.u.]')
    ax[2].set(title=f"{subject} - ECoG Decoding Accuracy")
    ax[2].set(xlim=(-6,6), ylim=(-6,6), xlabel='X-position [mm]', ylabel='Y-position [mm]')
    im.set_clim(0,1)
    cb.set_ticks(im.get_clim())
        
    fig.tight_layout()
    if save_figs:
        aopy.visualization.base.savefig(save_paths['ecog_maps'], f'{subject}_ecog_decoding_accuracy_maps_site_label.svg')
```

# Interpolate

```python
# Interpolate decoding accuracy map - nans first
for isubject, subject in enumerate(subjects):
    for iinsert, insert in enumerate(np_inserts):
        dec_map_results[subject][f"{insert}_interp"] = scipy.interpolate.interpn((np.unique(dec_map_results[subject]['data_map_ecogch_xy'][0]),np.unique(dec_map_results[subject]['data_map_ecogch_xy'][1])), dec_map_results[subject]['data_map_ecogch'].T, np_ecog_mapping[subject][insert]['pos'], method='linear', bounds_error=False)
```

```python
for isubject, subject in enumerate(subjects):
    fig, ax = plt.subplots(1,ninserts,figsize=(5*ninserts,4))
    for iinsert, insert in enumerate(np_inserts):
        npinsert_interp_datamap = aopy.visualization.get_data_map(dec_map_results[subject][f"{insert}_interp"], np_ecog_mapping[subject][insert]['pos'][:,0], np_ecog_mapping[subject][insert]['pos'][:,1])
        im = aopy.visualization.plot_spatial_map(npinsert_interp_datamap, np_ecog_mapping[subject][insert]['pos'][:,0], np_ecog_mapping[subject][insert]['pos'][:,1], cmap='YlGnBu', ax=ax[iinsert], nan_color='white')
        cb = plt.colorbar(im)
        ax[iinsert].set(title=f"ECoG Dec Acc: {subject} - {insert}")
        [ax[iinsert].plot(np_ecog_mapping[subject][f"{site_insert_name[subject][isite]}"]['pos'][site,0],np_ecog_mapping[subject][f"{site_insert_name[subject][isite]}"]['pos'][site,1], '.', color=dec_map_results[subject]['day_colors'][isite], markersize=10) for isite, site in enumerate(site_numbers[subject])]
    
    fig.tight_layout()
    plt.show()
```
